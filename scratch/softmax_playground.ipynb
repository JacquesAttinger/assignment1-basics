# Softmax playground

**TLDR:** Edit the score vector below, then run the cells to see how softmax turns scores into probabilities that add up to 1.

Softmax gives higher scores more probability. The function subtracts the largest score first, which keeps the calculation stable for large values.

```python
# Last edited: 2026-11-05 17:49 UTC
import numpy as np

import matplotlib.pyplot as plt

def softmax(scores):
    """Return the softmax probabilities for a one-dimensional score vector."""
    scores = np.asarray(scores, dtype=float)
    if scores.ndim != 1 or scores.size == 0:
        raise ValueError("scores must be a non-empty, one-dimensional vector")
    shifted_scores = scores - np.max(scores)
    exp_scores = np.exp(shifted_scores)
    return exp_scores / np.sum(exp_scores)
```

```python
# Change these scores and rerun this cell to explore different distributions.
scores = np.array([0.1, 0.1, 0.2])
probabilities = softmax(scores)

print("Scores:       ", scores)
print("Probabilities:", np.round(probabilities, 4))
print("Sum:          ", probabilities.sum())
```

```python
plt.figure(figsize=(7, 4))
bars = plt.bar(range(len(scores)), probabilities, color="#4C78A8")
plt.xticks(range(len(scores)), [f"Score {i}: {score:g}" for i, score in enumerate(scores)])
plt.ylim(0, 1)
plt.ylabel("Probability")
plt.title("Softmax distribution")
plt.bar_label(bars, fmt="%.3f", padding=3)
plt.tight_layout()
plt.show()
```
